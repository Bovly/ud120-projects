As you saw a little while ago, not every POI has an entry in the dataset (e.g. Michael Krautz). That’s because the dataset was created using the financial data you can find in final_project/enron61702insiderpay.pdf, which is missing some POI’s (those absences propagated through to the final dataset). On the other hand, for many of these “missing” POI’s, we do have emails.

While it would be straightforward to add these POI’s and their email information to the E+F dataset, and just put “NaN” for their financial information, this could introduce a subtle problem. You will walk through that here.

**How many people in the E+F dataset (as it currently exists) have “NaN” for their total payments? What percentage of people in the dataset as a whole is this?**

```python
unknown_total_payments = df['total_payments'].isnull().sum()
percentage_unknown_total_payments = (unknown_total_payments / df.shape[0]) * 100
print("Number of people with 'NaN' for total payments:", unknown_total_payments)
print("Percentage of people with 'NaN' for total payments:", percentage_unknown_total_payments,"%")
```

**How many POIs in the E+F dataset have “NaN” for their total payments? What percentage of POI’s as a whole is this?**

```python
poi_with_nan_total_payments = df[(df['poi'] == True) & (df['total_payments'].isnull())].shape[0]
total_pois = df['poi'].sum()
percentage_poi_nan_total_payments = (poi_with_nan_total_payments / total_pois) * 100

print("Number of POIs with 'NaN' for total payments:", poi_with_nan_total_payments)
print("Percentage of POIs with 'NaN' for total payments:", percentage_poi_nan_total_payments, "%")
```

If you added in, say, 10 more data points which were all POI’s, and put “NaN” for the total payments for those folks, the numbers you just calculated would change.

**What is the new number of people of the dataset? What is the new number of folks with “NaN” for total payments?**

```python
nan_mask = df['total_payments'].isna()

current_people = len(df)
current_nan_total_payments = int(nan_mask.sum())
current_POIs = int((df['poi'] == True).sum())

people_new = current_people + 10
nan_total_payments_new = current_nan_total_payments + 10
POIs_new = current_POIs + 10

people_new, nan_total_payments_new, POIs_new
print("New number of people in the dataset:", people_new
    ,"\nNew number of people with 'NaN' for total payments:", nan_total_payments_new
    ,"\nNew number of POIs in the dataset:", POIs_new
    )
```

**What is the new number of POI’s in the dataset? What is the new number of POI’s with NaN for total_payments?**

```python
POIs_nan_total_payments = int(((df['poi'] == True) & nan_mask).sum()) + 10

POIs_new, POIs_nan_total_payments
print("New number of POIs in the dataset:", POIs_new
    ,"\nNew number of POIs with 'NaN' for total payments:", POIs_nan_total_payments
    )
```
